```python
import sys
from pathlib import Path

import requests

# Shared Nestlé credentials / URL from project-root config (.env)
sys.path.insert(0, str(Path("..").resolve()))
from config import auth_headers, url

def summarize_markdown(markdown_content):
    request_payload = {
        "messages": [
            {
                "role": "system",
                "content": """You are a very helpful summarizer assistant that does exactly what it has been instructed. """
            },
            {
                "role": "user",
                "content": """Summarize the following markdown content.
                The size of the summary should be at least 60% the size of the original content.
                The resulting summary should also be in markdown format.
                Markdown content: \n"""+markdown_content
            }
        ]
    }

    req = requests.post(url, headers=auth_headers, json=request_payload, verify=False)
    res = req.json()
    return res["choices"][0]["message"]["content"]

def structure_correction(markdown_content):
    request_payload = {
        "messages": [
            {
                "role": "system",
                "content": """ """
            },
            {
                "role": "user",
                "content": """Correct the structure of this table markdown please:
                \n"""+markdown_content
            }
        ]
    }

    req = requests.post(url, headers=auth_headers, json=request_payload, verify=False)
    res = req.json()
    return res["choices"][0]["message"]["content"]
```

```python
import re
import pandas as pd
import numpy as np

def clean_markdown_table(raw_markdown):
    # Split the raw markdown by lines
    lines = raw_markdown.strip().split('\n')

    # Process the lines to split the columns
    table = [line.split('|')[1:-1] for line in lines]

    # Remove leading and trailing whitespaces from each cell
    table = [[cell.strip() for cell in row] for row in table]

    # Create a DataFrame
    df = pd.DataFrame(table[2:], columns=table[0])

    # Fill missing values in columns downwards
    for column in df.columns:
        df[column] = df[column].replace('', np.nan).ffill()

    return df.to_markdown(index=False)

import pandas as pd
from io import StringIO

def prepro(mdata):
    # Read the data into a DataFrame
    df = pd.read_csv(StringIO(mdata), sep="|", engine='python', on_bad_lines='warn')

    # Remove leading and trailing whitespace from column names
    df.columns = df.columns.str.strip()

    # Remove leading and trailing whitespace from each cell
    df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

    # Remove leading and trailing empty columns
    df = df.iloc[:, 1:-1]

    # Fill empty cells in the first column with the next non-empty cell
    for i in range(len(df) - 2, -1, -1):  # Start from second last row to the top
        if pd.isna(df.iloc[i, 0]) or df.iloc[i, 0] == '':
            df.iloc[i, 0] = df.iloc[i + 1, 0]

    # Replace NA with empty string for markdown conversion
    df = df.fillna('')

    # Output the corrected table
    corrected_markdown = df.to_markdown(index=False)
    return corrected_markdown

import pandas as pd
from io import StringIO


def prepro_2(mdata):
    # Read the data into a DataFrame
    df = pd.read_csv(StringIO(mdata), sep="|", engine='python')

    # Remove leading and trailing whitespace from column names
    df.columns = df.columns.str.strip()

    # Remove leading and trailing whitespace from each cell
    df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

    # Remove leading and trailing empty columns
    df = df.iloc[:, 1:-1]

    # Function to shift row elements to the right based on leading empty cells
    def shift_row(row):
        empty_count = (row == '').sum()
        if empty_count > 0:
            new_row = [''] * empty_count + row.dropna().tolist()
            return new_row[:len(row)]
        return row

    # Apply the function to each row
    shifted_df = df.apply(shift_row, axis=1, result_type='expand')

    # Replace NA with empty string for markdown conversion
    shifted_df = shifted_df.fillna('')

    # Output the corrected table
    corrected_markdown = shifted_df.to_markdown(index=False)
    return corrected_markdown
```

```python
import pandas as pd
import re
from openpyxl.utils import get_column_letter
from openpyxl import load_workbook

def markdown_table_to_excel(md_file_path, excel_file_path):
    with open(md_file_path, 'r') as file:
        content = file.read()

    tables = re.findall(r'\|(?:.*?\|)+', content, re.DOTALL)
    
    # Process each table
    for i, table in enumerate(tables):
        rows = table.split('\n')
        
        columns = [col.strip() for col in rows[0].split('|') if col.strip()]

        data = []
        for row in rows[2:]:
            if row.strip():
                data.append([cell.strip() for cell in row.split('|') if cell.strip()])

        df = pd.DataFrame(data, columns=columns)
        
        # Write to Excel file
        with pd.ExcelWriter(excel_file_path, engine='openpyxl', mode='a' if i > 0 else 'w') as writer:
            df.to_excel(writer, sheet_name=f'Table_{i+1}', index=False)

    # Load the workbook to adjust column widths
    wb = load_workbook(excel_file_path)
    for sheet in wb.sheetnames:
        ws = wb[sheet]
        for col in ws.columns:
            max_length = 0
            column = col[0].column_letter  # Get the column name
            for cell in col:
                try:
                    if len(str(cell.value)) > max_length:
                        max_length = len(cell.value)
                except:
                    pass
            adjusted_width = (max_length + 2)
            ws.column_dimensions[column].width = adjusted_width
    wb.save(excel_file_path)


markdown_table_to_excel('/nihs/NR/DSS/Project/WONI_DUND_102208/WONI_Zied/project/investigation_codes/tables_output/tables_markdown.md', 'output.xlsx')
```
